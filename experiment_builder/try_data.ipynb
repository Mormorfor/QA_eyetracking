```python
NEW_OUT_PATH = DATA_DIR / 'Experiment' / 'fake results' / 'Diana2'
```

```python
# The EyeLink Data Viewer ".xls" reports are actually UTF-16 (LE, BOM),
# tab-separated text files -> load and resave them as UTF-8 TSVs.
for xls_path in sorted(NEW_OUT_PATH.glob("*.xls")):
    df = pd.read_csv(
        xls_path,
        encoding="utf-16",
        sep="\t",
        quoting=csv.QUOTE_NONE,
        engine="python",
    )
    tsv_path = xls_path.with_suffix(".tsv")
    df.to_csv(tsv_path, sep="\t", index=False, encoding="utf-8")
    print(f"{xls_path.name}: {df.shape} -> {tsv_path.name}")
```
