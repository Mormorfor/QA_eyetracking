```python
from pathlib import Path
import sys
import os

PROJECT_ROOT = Path.cwd().parent

import pandas as pd
```

```python
from src.data_paths import (
    GATHERERS_PROCESSED_PATH,
    HUNTERS_PROCESSED_PATH,
    DATA_DIR,
)
```

```python
hunters = pd.read_csv(HUNTERS_PROCESSED_PATH)
gatherers = pd.read_csv(GATHERERS_PROCESSED_PATH)
all_participants = pd.concat([hunters, gatherers], ignore_index=True)
```

```python
all_participants
```

```python
text_cols = ["paragraph", "question", "answer_A", "answer_B", "answer_C", "answer_D"]

texts_per_qid = (
    all_participants.groupby("text_id_with_q")[text_cols].first().reset_index()
)
texts_per_qid
```

```python
rephrasings = pd.read_csv(DATA_DIR / "experiment_text - with rephrasings.csv", encoding="utf-8")
texts_per_qid["rephrased_answer_A"] = rephrasings["rephrased_answer_A"].values

out_path = DATA_DIR / "experiment_text_completed.csv"
texts_per_qid.to_csv(out_path, index=False)
print(f"Saved {len(texts_per_qid)} rows to {out_path}")
```

```python
from ftfy import fix_text
import pandas as pd

df = pd.read_csv(DATA_DIR / "experiment_text - with rephrasings.csv", encoding="latin1")

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].map(lambda x: fix_text(x) if isinstance(x, str) else x)

df.to_csv(DATA_DIR / "experiment_text - with rephrasings.csv", index=False, encoding="utf-8")
```
